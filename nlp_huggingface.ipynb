```python
from transformers import pipeline
```

```python
# classification
classifier=pipeline("sentiment-analysis")
classifier("Osama Bin Laden was right")
```

```python
classifier=pipeline("text-classification")
result=classifier("I like my college")
print(result)
```

```python
# name entity recognition
ner=pipeline("ner")
result=ner("Barack obama will visit India next year")
print(result)
```

```python
#question-answering
qa=pipeline("question-answering")
qa({"question":"Where is Ahad", "context":"Ahad went to the gym"})
```

```python
#text generation
generator=pipeline("text-generation")
generator("The future of AI is")
```

```python
#summarization
summarization=pipeline("summarization")
summarization(""" A file system is a method and data structure that an operating system uses to control how data is stored and retrieved on a storage device. The primary purpose of a file system is to manage the storage of data so that it is easy to find and access. It organises data into files and directories, maintains metadata, and manages the allocation of space on the storage medium. A file system is a crucial component of any operating system, serving as a framework for organising and managing the storage and retrieval of data on storage devices such as hard drives, SSDs, and USB drives. It provides a way to store data in an organised manner using files and directories, making it easy to locate and manipulate.
 """)
```

```python
#translation
translator=pipeline("translation", model="Helsinki-NLP/opus-mt-en-fr")
translator("yes")
```

```python
#masking
unmasker=pipeline("fill-mask")
unmasker("This course will teach <mask> language",top_k=4)
```

```python
zero_shot = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")

result = zero_shot(
    "This movie was a thrilling experience!",
    candidate_labels=["positive", "negative", "neutral"]
)
print(result)
```
